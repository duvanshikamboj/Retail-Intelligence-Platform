# Business Question 5

## What is the Average Order Value (AOV)?

### Objective

The objective of this analysis is to calculate the **Average Order Value (AOV)**, which represents the average amount of money spent by customers in a single order.

Average Order Value is an important business metric that helps evaluate customer purchasing behavior and revenue generated per transaction.

### Theory

Average Order Value (AOV) is calculated by dividing the total sales by the total number of unique orders.

### Formula


{Average Order Value} = {Total Sales}/{Total Orders}


A higher AOV indicates that customers spend more money per order, whereas a lower AOV suggests that customers make smaller purchases.

````python
average_order_value = total_sales / total_orders

print(f"Average Order Value : ${average_order_value:,.2f}")
````

### Observation

The Average Order Value (AOV) is **$491.39**.

This means that, on average, each customer order contributes **$491.39** in sales revenue.

### Business Insight

Average Order Value is an important indicator of customer spending behavior.

A higher AOV suggests that customers tend to purchase more expensive products or multiple products in a single transaction, leading to higher revenue per order.

### Recommendation

The company can increase the Average Order Value by implementing strategies such as:

- Product bundling
- Cross-selling related products
- Upselling premium products
- Offering free shipping above a minimum purchase amount

### Business Impact

Monitoring Average Order Value helps the company evaluate customer purchasing patterns and design pricing and marketing strategies that maximize revenue without necessarily increasing the number of customers.

# Business Question 14

## Which sub-categories require immediate business attention?

### Objective

The objective of this analysis is to identify the sub-categories that require immediate managerial attention due to poor financial performance.

Instead of focusing only on high-performing products, businesses must also identify underperforming areas that reduce overall profitability.

### Theory

A sub-category may require business attention if it:

- Generates negative profit (loss-making).
- Generates high sales but low profit.
- Performs significantly worse than other sub-categories.

Identifying such sub-categories helps management improve pricing strategies, optimize inventory, reduce unnecessary costs, and make informed business decisions.

````python
subcategory_performance = (
    df.groupby("Sub-Category")[["Sales", "Profit", "Quantity"]]
      .sum()
      .sort_values(by="Sales", ascending=False)
)

subcategory_performance
````

````python
def business_status(profit):
    if profit < 0:
        return "🔴 Needs Attention"
    elif profit < 50000:
        return "🟡 Moderate"
    else:
        return "🟢 Excellent"
````

````python
subcategory_performance["Status"] = (
    subcategory_performance["Profit"]
    .apply(business_status)
)
````

````python
subcategory_performance.head(10)
````

````python
top10 = subcategory_performance.head(10)

plt.figure(figsize=(12,6))

top10[["Sales","Profit"]].plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Top 10 Sub-Categories : Sales vs Profit")

plt.xlabel("Sub-Category")

plt.ylabel("Amount")

plt.xticks(rotation=45)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
````

## Business Performance Matrix

The following table summarizes the overall business performance of each sub-category based on Sales, Profit, Quantity, and Business Status.

This matrix helps management quickly identify:

- High-performing sub-categories
- Average-performing sub-categories
- Sub-categories requiring immediate business attention

It serves as a quick decision-making tool for inventory planning, pricing strategy, and resource allocation.

### Observation

The Business Performance Matrix provides a consolidated view of each sub-category's contribution.

It highlights that although some sub-categories generate strong sales, they do not necessarily contribute positively to profitability.

Sub-categories marked as **Needs Attention** should be investigated further to identify the root causes of poor financial performance.

### Business Insight

Rather than evaluating Sales or Profit independently, businesses should analyze multiple performance indicators together.

The Business Performance Matrix enables management to prioritize strategic actions by combining Sales, Profit, Quantity, and Business Status into a single decision-support framework.

### Recommendation

- Continue investing in sub-categories marked as **Excellent**.
- Monitor the performance of **Moderate** sub-categories and explore opportunities to improve profitability.
- Conduct a detailed root cause analysis for **Needs Attention** sub-categories before making pricing, inventory, or discontinuation decisions.

# Business Question 17

## Which customers are repeat customers?

### Objective

The objective of this analysis is to identify customers who have placed multiple orders with the company.

Identifying repeat customers helps businesses understand customer loyalty, improve customer retention strategies, and increase long-term revenue.

### Theory

Customer retention is one of the most important performance indicators in retail businesses.

A repeat customer is someone who has placed more than one order with the company. These customers generally contribute higher lifetime value and are less expensive to retain than acquiring new customers.

By counting the number of orders placed by each customer, we can identify loyal customers and understand purchasing behavior.

````python
repeat_customers = (
    df.groupby("Customer Name")["Order ID"]
      .nunique()
      .sort_values(ascending=False)
)

repeat_customers.head(10)
````

# Business Question 27

## Which States Require Immediate Business Attention?

### Objective

The objective of this analysis is to evaluate the overall business performance of each state by considering multiple performance indicators instead of relying only on sales or profit.

This analysis helps management quickly identify states that are performing well, states with average performance, and states requiring immediate business attention.

### Theory

Evaluating business performance using a single metric can lead to incorrect conclusions.

For example, a state may generate high sales but very low profit due to excessive discounts or high operational costs.

Therefore, multiple business metrics are combined:

- Total Sales
- Total Profit
- Profit Margin

Based on these metrics, every state is assigned a Business Status.

This approach provides a more comprehensive evaluation of regional business performance and supports strategic decision-making.

````python
state_performance = (
    df.groupby("State")[["Sales", "Profit"]]
      .sum()
      .reset_index()
)

state_performance.head()
````

````python
state_performance["Profit Margin (%)"] = (
    state_performance["Profit"] /
    state_performance["Sales"]
) * 100
````

````python
def business_status(margin):
    if margin >= 20:
        return "Excellent"
    elif margin >= 10:
        return "Moderate"
    else:
        return "Needs Attention"
````

````python
state_performance["Business Status"] = (
    state_performance["Profit Margin (%)"]
    .apply(business_status)
)

state_performance.head()
````

````python
state_performance = state_performance.sort_values(
    by="Profit Margin (%)",
    ascending=False
)

state_performance
````

````python
state_performance.head(15).style.background_gradient(
    subset=["Sales","Profit","Profit Margin (%)"],
    cmap="YlGn"
)
````

````python
state_performance.tail(15).style.background_gradient(
    subset=["Sales","Profit","Profit Margin (%)"],
    cmap="Reds"
)
````

````python
top15_margin = state_performance.nlargest(15, "Profit Margin (%)")

plt.figure(figsize=(12,6))

plt.barh(
    top15_margin["State"],
    top15_margin["Profit Margin (%)"]
)

plt.title("Top 15 States by Profit Margin")
plt.xlabel("Profit Margin (%)")
plt.ylabel("State")

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.show()
````

````python
bottom15_margin = state_performance.nsmallest(15, "Profit Margin (%)")

plt.figure(figsize=(12,6))

plt.barh(
    bottom15_margin["State"],
    bottom15_margin["Profit Margin (%)"]
)

plt.title("Bottom 15 States by Profit Margin")
plt.xlabel("Profit Margin (%)")
plt.ylabel("State")

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.show()
````

### Observation

The State Performance Matrix provides a consolidated view of sales, profit, and profit margin for every state.

States have been classified into three business categories based on their Profit Margin.

### Key Findings

- States with high profit margins are financially healthy and should be prioritized for future investments.
- Moderate-performing states have growth potential with suitable business strategies.
- States classified as "Needs Attention" require immediate investigation and corrective actions.

### Business Insight

Profit Margin is a better indicator of business efficiency than Sales alone.

A state generating high sales but low profit margin may not be financially sustainable in the long run.

The State Performance Matrix enables management to evaluate regional performance using multiple business indicators rather than relying on a single metric.

### Recommendation

- Expand operations in Excellent-performing states.
- Improve pricing strategies in Moderate-performing states.
- Review discount policies, logistics, and operational costs in states requiring attention.
- Monitor state-wise business performance regularly using Profit Margin.

### Business Impact

The State Performance Matrix provides a decision-support framework for regional business planning, resource allocation, profitability improvement, and long-term strategic growth.

# Business Question 33

## Which Month Performs the Best and Which Performs the Worst?

### Objective

The objective of this analysis is to identify the months that generate the highest and lowest sales.

Understanding the best and worst performing months helps businesses optimize inventory planning, marketing campaigns, staffing, and demand forecasting.

### Theory

Customer purchasing behavior changes throughout the year due to festivals, holidays, seasonal demand, promotional campaigns, and economic factors.

By analyzing monthly sales, businesses can identify recurring patterns and prepare strategies to maximize revenue during peak periods while improving performance during low-demand months.

````python
df["Month"] = df["Order Date"].dt.month_name()
````

````python
month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

df["Month"] = pd.Categorical(
    df["Month"],
    categories=month_order,
    ordered=True
)
````

````python
monthly_sales_summary = (
    df.groupby("Month")["Sales"]
      .sum()
      .sort_index()
)

monthly_sales_summary
````

````python
print("Best Month :")
display(monthly_sales_summary.nlargest(1))

print("\nWorst Month :")
display(monthly_sales_summary.nsmallest(1))
````

````python
plt.figure(figsize=(12,6))

monthly_sales_summary.plot(
    kind="bar",
    color="cornflowerblue"
)

plt.title("Monthly Sales Comparison")

plt.xlabel("Month")

plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
````

### Observation

The monthly sales comparison identifies the best and worst performing months based on total sales.

The variation in sales across months indicates seasonal customer purchasing behavior and changing market demand.

### Key Findings

- Certain months consistently generate higher sales.
- Some months contribute comparatively lower revenue.
- Monthly demand follows seasonal business patterns.

### Business Insight

Knowing the best-performing month enables businesses to prepare sufficient inventory, increase staffing, and launch premium marketing campaigns.

Similarly, understanding the worst-performing month helps management introduce promotional offers, discounts, and customer engagement strategies to improve sales.

### Recommendation

- Increase inventory before peak-demand months.
- Launch promotional campaigns during low-performing months.
- Allocate marketing budgets based on seasonal demand.
- Use monthly trends for demand forecasting and inventory optimization.

### Business Impact

Monthly performance analysis improves inventory planning, resource allocation, marketing effectiveness, and long-term business strategy.
