## What is Data Cleaning?

Data Cleaning is the process of identifying and correcting inaccurate, inconsistent, duplicate, incomplete, or improperly formatted data.

Real-world datasets are rarely perfect.

Before analysis or machine learning, a Data Scientist ensures that the dataset is clean and reliable.

Common data cleaning tasks include:

- Handling missing values
- Removing duplicate records
- Correcting data types
- Detecting outliers
- Standardizing inconsistent values
- Validating data integrity

A well-cleaned dataset improves the quality of business insights and predictive models.

```python
import pandas as pd
import numpy as np

# 1. Load Data
df = pd.read_csv("../data/raw/Global_Superstore_2016.csv", encoding="latin1")

# 2. Drop high null column
if "Postal Code" in df.columns:
    df.drop(columns=["Postal Code"], inplace=True)

# 3. Standardize Column Names (snake_case)
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_')

# 4. Fix Datatypes
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])

# 5. Sanity Check 1: Strip extra spaces from string columns
str_cols = df.select_dtypes(include=['object', 'string']).columns
for col in str_cols:
    df[col] = df[col].astype(str).str.strip()

# 6. Sanity Check 2: Shipping Days Feature (Model ke liye helpful hoga)
df['shipping_days'] = (df['ship_date'] - df['order_date']).dt.days

# Filter invalid transactions (agar shipping days negative hain)
df = df[df['shipping_days'] >= 0]

# 7. Sanity Check 3: Quantity aur Sales positive honi chahiye
df = df[(df['sales'] > 0) & (df['quantity'] > 0)]

# Summary Verification
print("--- CLEAN DATA SUMMARY ---")
print("Shape:", df.shape)
print("Null Values Remaining:", df.isnull().sum().sum())
print("Date Range:", df['order_date'].min(), "to", df['order_date'].max())
print("\nSample Clean Data:")
print(df.head(2))

# 8. Save CSV Correctly (without 'df =')
df.to_csv("../data/processed/cleaned_superstore.csv", index=False)
print("\nSuccessfully saved to ../data/processed/cleaned_superstore.csv")
```
