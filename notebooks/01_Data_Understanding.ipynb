#  Why Data Understanding?

Data Understanding is the first practical step after collecting a dataset.

Before applying any machine learning algorithm, a Data Scientist must understand:

- What information is available?
- What each row represents.
- What each column represents.
- Which columns are numerical and categorical.
- Whether there are missing values.
- Whether the data contains duplicate records.
- Whether the data types are correct.

Without understanding the dataset, any analysis or prediction may produce incorrect business decisions.

#### Step 3: Printing the first 5 rows; Why do we use df.head()?

The "head()" function displays the first five rows of the dataset.

Using this function, we can:

#### Step 8 : Null / Missing Values

Missing values represent unavailable or incomplete information.

If missing values are not handled properly, they may reduce model performance and produce inaccurate results.

Therefore, checking missing values is an essential part of Data Cleaning.

---

```python
df.isnull().sum()
```

#### Step 9 : Checking Duplicated values 

Duplicate records represent repeated observations in the dataset.

Duplicates may occur because of:

- Manual data entry errors
- System errors
- Data merging issues

Duplicate records can affect statistical analysis and machine learning performance.

Therefore, they should always be identified before analysis.

---

```python
df.duplicated().sum()
```

#### Step 10 : Checking data types

Every column in a dataset has a specific data type.

Examples include:

- Integer
- Float
- Object (Text)
- Boolean
- Datetime

Correct data types improve memory efficiency and ensure that analysis and machine learning algorithms work correctly.

Incorrect data types should be converted during the Data Cleaning stage.

---

```python
df.dtypes
```
